```python
import pandas as pd  
import csv
from sklearn.model_selection import train_test_split # ფუქნცია რომელიც მონაცემებს დაჰყოფს train/test-ად.
pd.set_option("display.max_colwidth", None)
```

```python
df = pd.read_csv(
    "../data/SMSSpamCollection",  
    sep="\t",                 
    header=None,              
    names=["label", "text"],  
    quoting=csv.QUOTE_NONE,    
)
```

```python
df.shape # rows and columns
```

```python
df.head()
```

```python
df["label"].value_counts()
```

```python
df["label"].value_counts(normalize=True)
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()

df = df.reset_index(drop=True)

df.shape
```

```python
df["label"].value_counts(normalize=True)
```

```python
df["length"] = df["text"].str.len()

df.groupby("label")["length"].mean()
```

```python
df["y"] = df["label"].map({"ham": 0, "spam" : 1})

df.head()
```

```python
X = df["text"]
y = df["y"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42,
)

len(X_train),len(X_test)
```

```python
print("Train:")
print(y_train.value_counts(normalize=True))
print()
print("Test:")
print(y_test.value_counts(normalize=True))
```

```python
train_df = pd.DataFrame({"text": X_train, "y": y_train})
test_df = pd.DataFrame({"text": X_test, "y" : y_test})

is_spam = train_df["y"] == 1

train_df[is_spam].sample(20, random_state=42)
```

```python
train_df[~is_spam].sample(20, random_state=42)
```

```python
train_df.to_csv("../data/train.csv", index=False)
test_df.to_csv("../data/test.csv", index=False)
```
